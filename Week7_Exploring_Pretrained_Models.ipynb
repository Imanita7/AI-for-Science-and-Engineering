#Predicting students at risk of poor academic performance using ML
##Project Problem
Students experience academic difficulties without realizing that they're falling behing. This project investigates whether ML can be used to identify students who are at risk of poor academic performance using factors such as previous grades, study time, absences, past failures and other student-related characteristics.

##Intended AI/ML outcome
The intended outcome is to develop a ML classification system that predicts whether a student is at risk of poor academic performance or not. The system will classify students into 2 categories:
    - At risk
    - Not at risk
The goal is to identify potential academic hardships early enough in order to support appropriate academic intervention.

##Data required
The project requires student-related academic and behavioural information. Potentially useful variables include:
  - Previous academic grades
  - Study time
  - Number of absences
  - Number of previous class failures
  - Access to educational support
  - Family educational support
  - Participation in extracurricular activities
  - Internet access
  - Family relationship quality
  - Free time
  - Health status
  - Other demographic and educational factors
  
The target variable will be the student's final academic performance.

````python
!pip install -q ucimlrepo
````

````python
from ucimlrepo import fetch_ucirepo
import pandas as pd

#Fetch the student performance dataset
student_performance = fetch_ucirepo(id=320)

#Get the features and targets
x = student_performance.data.features
y = student_performance.data.targets

#Combine them into one dataframe
df = pd.concat([x,y], axis = 1)

#Display the first few rows
df.head()
````

##Sample of the dataset
The table below shows a small portion of the student performance dataset.

````python
df.head(10)
print("Dataset shape:", df.shape)
df.info()
````

##Features and target
###Potential input features
The ML model will use factors that may be available before the final academic outcome is known including:
  - Study time
  - Number of previous failures
  - Number of absences
  - Educational support
  - Family support
  - Previous-period grades
  - Internet access
  - Free time
  - Health status
  - School-related and demographic factors

###Target
The target variable will be created from the final grade (e.g G3)
    - G3 < 10 = At risk
    - GE >= 10 = Not at risk
  
Therefore, this is a binary classification problem.

````python
#Create the target variable
df["at_risk"] = (df["G3"] < 10).astype(int)

#Risk variables
df["risk_label"] = df["at_risk"].map({
    0: "Not at risk",
    1: "At risk"
})

#Display the results
df[["G3", "at_risk", "risk_label"]].head(10)
````

G3 is used only to create the target variable and is not used as an input feature.

## Project Overview
This notebook explores a pretrained machine learning model available on Hugging Face and evaluates its relevance to our ongoing Student Performance Prediction project.

## 1. Introduction to pre-trained models
A pretrained model is a machine learning model that has already been trained on a large amount of data before being made available for other users and applications.

For this assignment, I explored pretrained models on Hugging Face and selected *TabPFN* because it is designed for classification tasks involving structured tabular data. Furthermore, this particular model suits my Student Performance Prediction project because:


* My dataset contains rows and columns of student information.
* My project is a classification problem.
* The model can work with numerical and categorical features.
* The model provides predictions for categories such as At Risk and Not At Risk.


## 2. The selected pretrained model
**Model Name:** TabPFN v2

**Model Repository:** Prior-Labs/TabPFN-v2-clf

**Hugging Face Link:**
https://huggingface.co/Prior-Labs/TabPFN-v2-clf

**Official GitHub Repository:**
https://github.com/PriorLabs/TabPFN

TabPFN is a pretrained machine learning model designed for classification and regression tasks involving tabular data.

It uses a transformer-based architecture and has been pretrained on many synthetic datasets. This allows it to learn general patterns that can be useful when working with new tabular datasets.

I selected this model because my Student Performance Prediction project also uses structured data containing student characteristics, study habits and academic information.

## 3. What does the model do?

TabPFN is designed to make predictions using structured datasets arranged in rows and columns.

It can be used for:

- Classification: Predicting which category a data record belongs to.
- Regression: Predicting numerical values.

### Inputs

The model receives:
- Training data containing input features.
- Corresponding target labels for classification.
- New data records that need predictions.

For my project, the input features include study time, absences and previous academic performance.

### Outputs

For a classification task, TabPFN can produce:
- Predicted class labels.
- Probability scores showing how likely each class is.

In my project, the predicted classes would be:

- At Risk
- Not At Risk

The model uses patterns learned during pretraining together with the supplied training data to make predictions.

## 4. How is the model relevant to my project?
TabPFN is relevant to our project because it is designed to work with structured tabular datasets.

Our project uses student information to identify learners who may be at risk of poor academic performance.

The model could be useful in the following ways:

1. **Student classification:** It could classify students into At Risk and Not At Risk categories.

2. **Early identification:** It could help identify students who may need additional academic support.

3. **Model comparison:** Its predictions could be compared with those from Logistic Regression and Random Forest.

4. **Educational research:** It could support the analysis of relationships between student behaviour, study habits and academic performance.

Therefore, TabPFN provides a potentially useful alternative for exploring student performance classification.

## 5. How does the model differ from the model developed in this course?
During this course, we developed and evaluated Logistic Regression and Random Forest models, which require training using selected dataset to learn patterns and make predictions, to classify students as At Risk or Not At Risk.

TabPFN differs from these models because it is a pretrained foundation model that has already learned general patterns from previously generated tabular datasets. Therefore, it can use prior knowledge when working with a new dataset.

| Feature | Logistic Regression | Random Forest | TabPFN |
|---|---|---|---|
| Model type | Traditional machine learning | Ensemble learning | Pretrained transformer-based model |
| Training | Learns from project training data | Learns from project training data | Pretrained on synthetic tabular tasks and uses supplied training data |
| Main approach | Learns a relationship between features and classes | Combines predictions from multiple decision trees | Uses patterns learned during pretraining to make predictions |
| Application | Classification | Classification and regression | Classification and regression |

Unlike Logistic Regression and Random Forest, TabPFN has already learned general patterns during pretraining.

However, it still requires our project training data when making predictions for the student classification task.

Exploring TabPFN would allow us to understand how pretrained models can complement traditional machine learning approaches.

## 7. Limitations and precautions
Although TabPFN offers useful possibilities, there are limitations and precautions to consider.

1. **Data quality:** Inaccurate, incomplete or inconsistent student data may affect prediction results.

2. **Bias:** The model may produce biased predictions if the data used does not adequately represent different groups of students.

3. **Privacy:** Student information is sensitive and should be protected when using AI systems.

4. **Interpretability:** Understanding exactly why a model predicts that a student is at risk may be difficult.

5. **Computational requirements:** Running TabPFN may require sufficient memory and computational resources.

6. **Prediction accuracy:** A pretrained model does not guarantee better results than models trained specifically for our project.

7. **Human judgement:** Predictions should not be used as the only basis for making decisions about students.

The model should be evaluated using appropriate performance metrics, and its predictions should support rather than replace human judgement.
